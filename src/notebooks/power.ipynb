#### Compare the two $\alpha$'s

```python
def solve_for_alpha_det(u, v):
    def equation(alpha):
        return alpha + 1 - 2*(u + v) + v * (2.0 ** (-alpha))
    try:
        ans = fsolve(equation, v, xtol=1e-8)[0]
        if abs(equation(ans)) < 1e-5:
            return ans
    except:
        pass
    return np.nan

grid_resolution = 20 
u_vals = np.linspace(0, 1, grid_resolution)
v_vals = np.linspace(0, 1, grid_resolution)
U, V = np.meshgrid(u_vals, v_vals)

domain_mask = ((2 * U + V) > 1) & ((U + V) <= 1)
diff_grid = np.full_like(U, np.nan)

  
print("Computing grid... this may take a moment.")
for i in range(grid_resolution):
    for j in range(grid_resolution):
        if domain_mask[i, j]:
            print(f"Computing for u={U[i, j]:.2f}, v={V[i, j]:.2f}")
            a_det = solve_for_alpha_det(U[i, j], V[i, j])
            a_sto = find_alpha_mean_preserving(U[i, j], V[i, j])
            diff_grid[i, j] = a_sto - a_det


plt.figure(figsize=(9, 7))

plt.contourf(U, V, domain_mask, levels=[0.5, 1.5], colors=['#e0e0e0'], alpha=0.3)

# Plot the heatmap
cmap = plt.cm.coolwarm
# Using pcolormesh for grid-based heatmap rendering
mesh = plt.pcolormesh(U, V, diff_grid, shading='nearest', cmap=cmap)
cbar = plt.colorbar(mesh)
cbar.set_label(r'$\Delta \alpha = \alpha_{stoch} - \alpha_{det}$')

plt.xlabel('u')
plt.ylabel('v')
plt.title(r'Deviation between Stochastic and Analytical $\alpha$')
plt.xlim([0, 1])
plt.ylim([0, 1])
plt.grid(True, linestyle='--', alpha=0.4)

plt.show()
```
